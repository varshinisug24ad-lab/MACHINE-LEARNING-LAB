```python
from sklearn.datasets import fetch_20newsgroups
import pandas as pd

train_data = fetch_20newsgroups(
    subset="train",
    remove=("headers", "footers", "quotes")
)

test_data = fetch_20newsgroups(
    subset="test",
    remove=("headers", "footers", "quotes")
)

print("Train documents:", len(train_data.data))
print("Test documents:", len(test_data.data))
```

```python
train_df = pd.DataFrame({
    "documents": train_data.data,
    "labels": train_data.target
})

test_df = pd.DataFrame({
    "documents": test_data.data,
    "labels": test_data.target
})

train_df.to_csv("train_20newsgroups.csv", index=False)
test_df.to_csv("test_20newsgroups.csv", index=False)

print("Train Shape:", train_df.shape)
print("Test Shape:", test_df.shape)
```
